```python
import asyncio
import time
```

```python
# Sync timer
def count():
    print("One")
    time.sleep(1)
    print("Two")
    time.sleep(1)
    print("Three")
    time.sleep(1)

def call():
   for _ in range(3):
        count()

def main():
    start = time.perf_counter()
    call()
    end = time.perf_counter()-start
    print("Sync counter Time: ", end," seconds")
main()
```

```python
async def async_count():
    print("One")
    await asyncio.sleep(1)
    print("Two")
    await asyncio.sleep(1)
    print("Three")
    await asyncio.sleep(1)

async def async_call():
   await asyncio.gather(async_count(), async_count(), async_count())

async def async_main():
    start = time.perf_counter()
    await async_call()
    end = time.perf_counter()-start
    print("Async counter Time: ", end," seconds")

await async_main()
```

```python
asyncio.run(async_main()) # create event loop and run until complete , but it is only used in eventloops, notebook environment is already running an event loop

time.sleep(5) # time consuming blocking function call
asyncio.sleep(5) # non blocking function call
```
